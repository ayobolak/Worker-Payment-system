```python
import random
```

```python
# Step 1: Generate 400 workers

def generate_workers(num_workers=400):
    names = ["John", "Jane", "Mary", "Mike", "Anne", "Bob", "Rose", "Tom", "Lily", "Chris"]
    workers = []

    for i in range(num_workers):
        name = random.choice(names) + str(i)
        gender = random.choice(["Male", "Female"])
        salary = random.randint(5000, 35000)
        workers.append({"id": i + 1, "name": name, "gender": gender, "salary": salary})

    return workers
```

```python
# Step 2: Generate payment slips
def generate_payment_slips(workers):
    for worker in workers:
        emp_level = "Unclassified"
        salary = worker["salary"]
        gender = worker["gender"]

        if 10000 < salary < 20000:
            emp_level = "A1"
        if 7500 < salary < 30000 and gender == "Female":
            emp_level = "A5-F"

        print(f"Payment Slip - ID: {worker['id']}, Name: {worker['name']}, "
              f"Gender: {gender}, Salary: ${salary}, Level: {emp_level}")
```

```python
if __name__ == "__main__":
    workers_list = generate_workers()
    generate_payment_slips(workers_list)
```
